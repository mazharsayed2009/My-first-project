## For Loop Explained Simply

A `for` loop in Python is used to iterate over a sequence (like a list, tuple, dictionary, string, or `range`). It's perfect when you want to repeat a block of code a specific number of times, or for each item in a collection.

### The `range()` Function

A very common way to use `for` loops is with the `range()` function. This function generates a sequence of numbers, which the loop then iterates through.

`range()` can be used in a few ways:

1.  `range(stop)`: Generates numbers from 0 up to (but not including) `stop`.
    *   Example: `range(5)` will give `0, 1, 2, 3, 4`.

2.  `range(start, stop)`: Generates numbers from `start` up to (but not including) `stop`.
    *   Example: `range(2, 7)` will give `2, 3, 4, 5, 6`.

3.  `range(start, stop, step)`: Generates numbers from `start` up to (but not including) `stop`, incrementing by `step` each time.
    *   Example: `range(1, 10, 2)` will give `1, 3, 5, 7, 9`.

### Basic `for` Loop Structure with `range()`

```python
for variable in sequence:
    # code to be executed for each item in the sequence
    # 'variable' will take on the value of each item in turn
```

Let's look at some examples to make this concrete.

````python
# Example 1: Printing numbers from 0 to 4
print("Counting from 0 to 4:")
for num in range(5):
    print(num)

print("\n--- Output of next loop ---")
# Example 2: Printing numbers from 2 to 6
print("Counting from 2 to 6:")
for i in range(2, 7):
    print(i)

print("\n--- Output of next loop ---")
# Example 3: Printing odd numbers from 1 to 9
print("Counting odd numbers from 1 to 9:")
for j in range(1, 10, 2):
    print(j)
````

### `for` Loop with other sequences (like lists)

You can also iterate directly over items in a list, string, or other collection.

````python
# Example 4: Iterating through a list of fruits
fruits = ["apple", "banana", "cherry"]
print("My favorite fruits are:")
for fruit in fruits:
    print(fruit)

print("\n--- Output of next loop ---")
# Example 5: Iterating through characters in a string
word = "Python"
print("Letters in the word 'Python':")
for char in word:
    print(char)
````

### Key takeaways for `for` loops:

*   They are used for **iteration** (going through items one by one).
*   They work with any **sequence** (lists, strings, tuples, `range()`, etc.).
*   The `range()` function is often used to generate a **sequence of numbers** for a loop.
*   The `variable` in the `for` statement takes on the **value of the current item** in each iteration.

**Break** **and** **Continue** **Statement**

### `break` Statement 🛑

Imagine you're playing a game (your loop). `break` is like pressing the 'Exit Game' button. Jaise hi `break` aata hai, loop wahin ruk jaata hai, aur baaki ki cheezein ignore ho jaati hain. Bas, game over for that loop!

**Jab use karte hain:** Jab tumhe ek specific condition poori hone par loop se *poori tarah bahar nikalna* ho.

````python
# Example for `break`
print("Counting till 5 (using break at 5):")
for i in range(10): # Ye loop 0 se 9 tak chalna chahiye
    if i == 5: # Lekin jaise hi i, 5 hota hai...
        print("Number 5 aa gaya! Break kar rahe hain! 👋")
        break # ...loop wahin ruk jayega
    print(f"Current number: {i}")

print("Loop khatam (break ki wajah se)!")
````

### `continue` Statement ➡️

Ab imagine karo, wahi game chal rahi hai. `continue` is like pressing 'Skip Level' button for *this one round*. Current round skip ho jaati hai, uske baad ki cheezein execute nahi hoti, but loop next round se dobara shuru ho jaata hai. Tum game se bahar nahi nikalte, bas current level chhod dete ho.

**Jab use karte hain:** Jab tumhe ek specific condition poori hone par current iteration ko *skip karke agle iteration pe jaana* ho.

````python
# Example for `continue`
print("Printing numbers, but skipping 3:")
for i in range(5): # Ye loop 0 se 4 tak chalna chahiye
    if i == 3: # Jaise hi i, 3 hota hai...
        print("Number 3 hai, skip kar rahe hain! 🏃‍♂️")
        continue # ...current iteration skip ho jayegi, aur agla number print hoga
    print(f"Current number: {i}")

print("Loop khatam!")
````

*   **`continue`**: ⏭️ Aage badho! Ye *current step skip karo* aur agla step shuru karo. Jaise tumne 'Skip' button dabaya.

First multiple found

Task: Given an array of integers, find and print the index of the first element that is a multiple of 7. If none, print -1. Use break once found.

Hint: Iterate with index; break as soon as you find arr[i] % 7 == 0.

````python
arr = [1, 5, 10, 14, 20, 21, 28, 35]
found_multiple = False

for i in range(len(arr)): # Array के हर element पर Iterate करो
    if arr[i] % 7 == 0: # Check करो कि क्या यह 7 का मल्टीपल है?
        print(f"पहला मल्टीपल index {i} पर मिला, जिसकी value है: {arr[i]}")
        found_multiple = True
        break # जैसे ही मिल जाए, लूप को रोक दो

if not found_multiple: # अगर लूप पूरा हो गया और मल्टीपल नहीं मिला
    print(-1)
````

````python
arr = [1,5,10,14,20,21,28,35]
found_multiple = False
for i in range(len(arr)):
  if arr[i] % 7 == 0:
    print(f"first multiple index {i} and value is {arr[i]}")
    found_multiple = True
    break
if not found_multiple:
  print(-1)
````
